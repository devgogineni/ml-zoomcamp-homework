```python
df_asia = df[df['origin'] == 'Asia']
df_asia.head()
```

```python
df_asia = df_asia[['vehicle_weight', 'model_year']].head(7 )
df_asia.head(7)
```

```python
X = df_asia[["vehicle_weight", "model_year"]].to_numpy(dtype=float)
X.shape
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200], dtype=float)
y
```

```python
print("y shape:", y.shape)  # Should be (7,)
```

```python
print(X.shape)
print(X.T.shape)
print(y.shape)
```

```python
w = XTX_inv @ X.T @ y
print(w)
```
